## Key Insights

1. Fraud cases represent a small minority of transactions.

2. Purchase values are right-skewed.

3. Customer ages are concentrated in a limited range.

4. Fraud patterns differ across browsers and traffic sources.

5. Dataset shows severe class imbalance requiring resampling.
