```python
import pandas as pd
```

```python
df = pd.read_csv(
    "../data/raw/Fraud_Data.csv"
)
```

```python
df["signup_time"] = pd.to_datetime(
    df["signup_time"]
)

df["purchase_time"] = pd.to_datetime(
    df["purchase_time"]
)
```

```python
df["time_since_signup"] = (
    df["purchase_time"]
    -
    df["signup_time"]
).dt.total_seconds()
```

```python
df["hour_of_day"] = (
    df["purchase_time"]
    .dt.hour
)
```

```python
df["day_of_week"] = (
    df["purchase_time"]
    .dt.dayofweek
)
```

```python
transaction_frequency = (
    df.groupby("user_id")
    .size()
    .reset_index(
        name="transaction_count"
    )
)

transaction_frequency.head()
```

## Resampling Strategy

The fraud datasets are highly imbalanced.

SMOTE will be applied during model training because it generates synthetic fraud examples and helps improve recall without discarding legitimate transactions.

Resampling will only be applied on training data to avoid data leakage.
